# Ensemble vs Time

```python
import numpy as np

import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
rng = np.random.default_rng(seed=42)
```

## Generation of an Ensemble of Realizations of a Stochastic Process

AR(1) — first-order autoregression:

$$
X_t = \varphi\cdot X_{t-1} + \varepsilon_t, \quad \varepsilon_t \sim \mathcal{N}(0, \sigma^2)
$$

```python
N_real = 50        # number of realization
T = 500            # length of each realization
phi = 0.8          # coefficient of AR(1)
sigma = 1.0        # white noise

# white noise
white = rng.normal(0, sigma, size=(N_real, T))

ar1 = np.zeros((N_real, T))
ar1[:, 0] = white[:, 0] # initial state (0 is okay too)
for t in range(1, T):
    ar1[:, t] = phi * ar1[:, t-1] + white[:, t]

ensemble = white + ar1
ensemble.shape # (N_real, T)
```

```python
plt.figure(figsize=(18, 4))
plt.plot(ensemble[:5].T, linewidth=.7)
plt.title('Realization AR(1) + white noise')
```
